```python
# ques 2

# Inventry Control System

# Task

# creat [5,5] matrix (stock level):
x = np.random.randint(20,81,(5,5))
print("\n stock level")
print(x)


# creat another 5*5 array (demand):
y = np.random.randint(20,81,(5,5))
print("\n demand")
print(y)

# remaining stock = stock - demand:
print("\n remaining stock")
print(x-y)

# negative values -> 0 out of stock
print("\n out of stock")
outOfStock = np.where(x-y<0,0,x-y)
print(outOfStock)

# total shortage
print("\n total shortage")
total_shortage = np.sum(np.where(x-y<0,(x-y)*-1,0),axis=0)
print(total_shortage)

# products with zero stock
zero_stock=np.sum(np.where(x-y==0,1,0),axis=0)
print("\n products with zero stock")
print(zero_stock)

# sort each row based on remaining stock
print("\n sort each row based on remaining stock")
print(np.sort(outOfStock,axis=1))

```

```python
# single inheritance
class person:
  def __init__(self,h,w,age):
    self.height = h
    self.weight = w
    self.age = age

class rahul(person):

  def __init__(self,h,w,age,name,g,b):

    super().__init__(h,w,age)
    self.gender=g
    self.name=name
    self.bank_balance=b


obj = rahul(6,80,18,'Rahul','male',18000)
print(obj.gender)
print(obj.name)
print(obj.bank_balance)
print(obj.height)
```

```python
class complex:
  def __init__(self,x,y):
    self.x = x
    self.y = y

  def __add__(self,other):
    return complex(self.x+other.x,self.y+other.y)

  # Renamed to 'display' and removed 'other' argument
  def display(self):
    print(self.x,"+",self.y,"i")

a = complex(2,3)
b = complex(4,5)
c = a+b
print(c.x)
print(c.y)
c.display()
```
