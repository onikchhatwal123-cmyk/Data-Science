```python
weekly_sales = {}

while True:
    item = input().strip().lower()

    if item == "end":
        break

    quantity = int(input(f"How many {item}s were sold?: "))

    if item in weekly_sales:
        weekly_sales[item] += quantity
    else:
        weekly_sales[item] = quantity
    print(f"Recorded: +{quantity} {item}(s)\n")


    for item, total_sold in weekly_sales.items():
        print(f"• {item.capitalize()}: {total_sold} sold")
```

```python
import numpy as np

x = np.random.randint(1,91,(5, 5))

row_mean = np.mean(x, axis=1,dtype=int)

column_mean = np.mean(x, axis=0,dtype=int)

modified_mat = np.where(x%2==0,-1,x)

sort_row = np.sort(x,axis=1)

sort_col = np.sort(x,axis=0)

sum_row = np.sum(x,axis=1)

sum_col = np.sum(x,axis=0)

count_col = np.sum(np.where(modified_mat<0,1,0),axis=0)

print("\n array")
print(x)

print("\n row mean")
print( row_mean)

print("\n col mean")
print(column_mean)

print("\n replace even no. with -1")
print(modified_mat)

print("\n row sort")
print(sort_row)

print("\n col sort")
print(sort_col)

print("\n row sum")
print(sum_row)

print("\n col sum")
print(sum_col)

print("\n count the number of -ves ")
print(count_col)
```

```python
# ques 2

# Inventry Control System

# Task

# creat [5,5] matrix (stock level):
x = np.random.randint(20,81,(5,5))
print("\n stock level")
print(x)


# creat another 5*5 array (demand):
y = np.random.randint(20,81,(5,5))
print("\n demand")
print(y)

# remaining stock = stock - demand:
print("\n remaining stock")
print(x-y)

# negative values -> 0 out of stock
print("\n out of stock")
outOfStock = np.where(x-y<0,0,x-y)
print(outOfStock)

# total shortage
print("\n total shortage")
total_shortage = np.sum(np.where(x-y<0,(x-y)*-1,0),axis=0)
print(total_shortage)

# products with zero stock
zero_stock=np.sum(np.where(x-y==0,1,0),axis=0)
print("\n products with zero stock")
print(zero_stock)

# sort each row based on remaining stock
print("\n sort each row based on remaining stock")
print(np.sort(outOfStock,axis=1))

```
